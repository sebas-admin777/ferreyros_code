Crear un DataFrame con Pandas.

```python
import pandas as pd
#Crear DataFrame
df =pd.DataFrame({
    "Producto":["A","B","C","D"],
    "Ventas": [100, 150, 80, 120],
    "Precio": [10, 12, 8, 9]
})
#columna calculada
df["Ingresos"] = df["Ventas"] * df["Precio"]

# Filtrar
print ("Productos con ventas > 100:\n", df[df["Ventas"] > 100])

# Ordenar
print("ordenando por ingresos:\n", df.sort_values(by="Ingresos", ascending=False))

#Estadisticas
print("Estadisticas:\n",df.describe())
```

```python
producto = pd.read_csv('S1-T3.csv', encoding='latin-1')
producto
```

Agregar la columna Ingresos y calcular los Ingresos por fila = Cantidad * Precio_unitario * (1 - Descuento).

```python
producto['Ingresos'] = producto['Cantidad'] * producto['Precio_unitario'] * (1 - producto['Descuento'])
producto
```


Muestra las estadísticas básicas

```python
print(producto.describe())
```

Ventas totales por producto (sumar Cantidad e Ingresos).

```python
ventas_producto = producto.groupby('Producto').agg(
  Total_Cantidad=('Cantidad', 'sum'),
  Total_Ingresos=('Ingresos', 'sum')
).reset_index().sort_values(by='Total_Ingresos', ascending=False)


ventas_producto
```

Análisis por categoría y por fecha

```python
# Por categoría

ventas_categoria = producto.groupby('Categoria').agg(
  Total_Cantidad=('Cantidad','sum'),
  Total_Ingresos=('Ingresos','sum')
).reset_index()


# Por fecha (serie temporal)
ventas_dia = producto.groupby('Fecha').agg(
  Ingresos_Diarios=('Ingresos','sum'),
  Cantidad_Diaria=('Cantidad','sum')
).reset_index()

ventas_categoria, ventas_dia
```

Crear un arreglo numérico (Cantidad, Precio_unitario, Ingresos) y hacer operaciones:

```python
import numpy as np
arreglo = producto[['Cantidad', 'Precio_unitario', 'Ingresos']].to_numpy() # ndarray n x 3
# Sumas y promedios
suma_total = np.sum(arreglo)       # suma de todos los valores numéricos en la submatriz
suma_por_col = np.sum(arreglo, axis=0)  # [sum Cantidad, sum Precio_unitario, sum Ingresos]
promedio_col = np.mean(arreglo, axis=0)

# Transposición
arreglo_T = arreglo.T
print("Arreglo: \n", arreglo)
print("Arreglo traspuesto : \n", arreglo_T)
print("Suma por columna:", suma_por_col)
print("Promedio por columna:", promedio_col)
print("Columnas y filas:", arreglo_T.shape)
```
